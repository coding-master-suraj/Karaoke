```python
# 1. Install Demucs AI Model (100% Free Studio Separation Engine)
!pip install -U demucs

from google.colab import files
import os

# 2. Upload your Song Track
print("Apnar Song File-ti Upload Korun:")
uploaded = files.upload()

for fn in uploaded.keys():
  input_file = fn

# 3. Separate Audio Stems using High Precision Demucs Model
print("AI Model is separating vocals and preserving original background music...")
!demucs -d cpu "{input_file}"

# 4. Download 100% Clean Instrumental Track
filename = os.path.splitext(input_file)[0]
output_path = f"separated/htdemucs/{filename}/no_vocals.wav"

if os.path.exists(output_path):
    print("Separation Complete! Downloading Instrumental File...")
    files.download(output_path)
else:
    print("Processing Completed! Check Output Folder.")
```
