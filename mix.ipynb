```python
# 1. Install Demucs AI Engine
!pip install -U demucs

import os
import shutil
import time
import gc
from google.colab import files

# --- প্রসেস শুরুর আগেই পুরোনো ক্যাশ ও ফাইল ডিলিট করার ফাংশন ---
def clean_previous_files():
    gc.collect()
    if os.path.exists("separated"):
        shutil.rmtree("separated", ignore_errors=True)
    # Colab ডিরেক্টরিতে থাকা পুরোনো অডিও ফাইলগুলো ডিলিট
    for file in os.listdir('.'):
        if file.endswith('.wav') or file.endswith('.mp3') or file.endswith('.flac'):
            try:
                os.remove(file)
            except Exception:
                pass

# ১. আগের সব ফাইল ডিলিট
clean_previous_files()

# ২. Upload Song
print("🎵 আপনার নতুন গানটি Upload করুন:")
uploaded = files.upload()

if not uploaded:
    print("❌ কোনো ফাইল আপলোড করা হয়নি!")
else:
    input_file = list(uploaded.keys())[0]
    filename = os.path.splitext(input_file)[0]

    # ৩. GPU দিয়ে 4-Stem AI Separation
    print("\n🚀 AI সেপারেশন শুরু হচ্ছে (Vocals, Drums, Bass, Other)...")
    !demucs -d cuda "{input_file}"

    # সেপারেট হওয়া ফাইলের লোকেশন
    stem_dir = f"separated/htdemucs/{filename}"
    stems = ["vocals.wav", "drums.wav", "bass.wav", "other.wav"]

    # ৪. প্রতিটি আলাদা ফাইল টাইম গ্যাপ দিয়ে ডাউনলোড
    print("\n✅ সেপারেশন সম্পন্ন হয়েছে! ফাইলগুলো পরপর ডাউনলোড শুরু হচ্ছে...\n")
    print("⚠️ ব্রাউজারে 'Allow multiple downloads' চাওয়া হলে 'Allow' এ ক্লিক করুন।\n")

    for idx, stem in enumerate(stems, start=1):
        file_path = os.path.join(stem_dir, stem)
        if os.path.exists(file_path):
            print(f"📥 [{idx}/4] Downloading: {stem} ...")
            files.download(file_path)
            # ব্রাউজার ব্লকিং এড়াতে ৩ সেকেন্ডের বিরাম (Pause)
            time.sleep(3)
        else:
            print(f"⚠️ পাওয়া যায়নি: {stem}")

    print("\n🎉 ৪টি ফাইলই ডাউনলোডের জন্য পাঠানো শেষ হয়েছে!")
```
